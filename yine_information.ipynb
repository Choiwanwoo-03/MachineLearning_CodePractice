### 스케일링 적용하기

```python
# 학습셋과 시험셋 분리
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(data.drop('class', axis = 1), data['class'],
                                                    test_size = 0.2, random_state = 100)
```

```python
# 최소-최대 스케일러를 사용하여 학습 및 변환
mm_scaler = MinMaxScaler()
X_train_scaled = mm_scaler.fit_transform(X_train)   # 학습셋 학습 및 트랜스폼
X_test_scaled = mm_scaler.transform(X_test)         # 시험셋 트랜스폼
```

```python
# 모델링 및 예측하기
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier()
knn.fit(X_train_scaled, y_train)
pred = knn.predict(X_test_scaled)
```

```python
pred
```

```python
# 평가하기
from sklearn.metrics import accuracy_score
accuracy_score(y_test, pred)
```

### 하이퍼파라미터 튜닝하기

```python
# n_neighbors 값 조절하기
knn = KNeighborsClassifier(n_neighbors = 7) # 이웃 7개
knn.fit(X_train_scaled, y_train)
pred = knn.predict(X_test_scaled)
accuracy_score(y_test, pred)
```

```python
knn = KNeighborsClassifier(n_neighbors = 3) # 이웃 3개
knn.fit(X_train_scaled, y_train)
pred = knn.predict(X_test_scaled)
accuracy_score(y_test, pred)
```

```python
# 반복 코드로 n_neighbors 값 조절하기
scores = []
for i in range(1, 21) :
    knn = KNeighborsClassifier(n_neighbors = i)
    knn.fit(X_train_scaled, y_train)
    pred = knn.predict(X_test_scaled)
    acc= accuracy_score(y_test, pred)
    scores.append(acc)
```

```python
# 20개의 값 점수 확인
scores
```

```python
# 라인 그래프로 예측력의 변화를 시각화하기
import seaborn as sns
sns.lineplot(x = range(1, 21), y = scores)
```

```python
# 최종 매개변수 : 13
knn = KNeighborsClassifier(n_neighbors = 13)
knn.fit(X_train_scaled, y_train)
pred = knn.predict(X_test_scaled)
accuracy_score(y_test, pred)
```
