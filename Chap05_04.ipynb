### 5.4 붓꽃 데이터를 이용한 k_NN 활용 연습

```python
from sklearn.datasets import load_iris

iris = load_iris()
iris.data[:5]
```

```python
print('iris 데이터의 형태 : , iris.data.shape')
print('iris 데이터의 속성들 : ', iris.feature_names)
print('iris 데이터의 레이블 : ', iris.target)
```

```python
import pandas as pd

iris_df = pd.DataFrame(iris.data, columns = iris.feature_names)
iris_df['target'] = pd.Series(iris.target)
iris_df.head()
```

```python
iris_df['target'].value_counts()
```

```python
iris_df.values
```

```python
X = iris_df.iloc[:, :4]
y = iris_df.iloc[:, -1]

print(X, y)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn import metrics

def iris_knn(X, y, k) :
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3)
    knn = KNeighborsClassifier(n_neighbors = k)
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    return metrics.accuracy_score(y_test, y_pred)

k = 3
scores = iris_knn(X, y, k)
print('n_neihghbors가 {0:d}일때 정확도 : {1:3f}'.format(k, scores))
```
