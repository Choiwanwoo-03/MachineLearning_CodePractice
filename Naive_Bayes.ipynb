### 문자 데이터셋을 이용해 스팸 여부를 판단하라.

데이터셋 : 'https://media.githubbusercontent.com/media/musthave-ML10/data_source/main/spam.csv'

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

file_url = 'https://media.githubusercontent.com/media/musthave-ML10/data_source/main/spam.csv'
data = pd.read_csv(file_url)
```

```python
import string
```

```python
string.punctuation
```

```python
sample_string = data['text'].loc[0]
sample_string
```

```python
# 문자열의 문자를 하나씩 출력
for i in sample_string :
    print(i)
```

```python
# 특수 기호가 아닌 문자만 보이게 하기
for i in sample_string :
    if i not in string.punctuation :
        print(i)
```

```python
# 특수 기호를 제외한 문자들을 리스트에 모아주기
new_string = []
for i in sample_string:
  if i not in string.punctuation:
    new_string.append(i)
```

```python
new_string
```

```python
# 리스트 안의 문자들을 문자열로 만들기
new_string = ''.join(new_string)
new_string
```

```python
# 종합해서 함수로 만들기
def remove_punc(x):
  new_string = []
  for i in x:
    if i not in string.punctuation:
      new_string.append(i)

  new_string = ''.join(new_string)

  return new_string
```

```python
# apply() 함수로 한 행마다 특수기호 제거하기
data['text'] = data['text'].apply(remove_punc)
data['text']
```

```python
# nltk 라이브러리에서 불용어 목록 가져오기
import nltk
nltk.download('stopwords')
```

```python
# stopword 임포트 하기
from nltk.corpus import stopwords
```

```python
# stopwords에서 제공하는 리스트 확인
stopwords.words('english')
```

```python
# 단어 단위로 문장 분할
sample_string = data['text'][0]
sample_string.split()
```

```python
# 각 단어가 불용어에 속하는 지 아닌지 판독
for i in sample_string.split() :
    if i.lower() not in stopwords.words('english') :
        print(i.lower())
```

```python
# 불용어 제거 함수 만들기
def stop_words(x) :
    new_string = []
    for i in x.split() :
        if i.lower() not in stopwords.words('english') :
            new_string.append(i.lower())
    new_string = ' '.join(new_string)
        
    return new_string
```

```python
# apply() 함수로 한 행마다 불용어 제거하기
data['text'] = data['text'].apply(stop_words)
data['text']
```
