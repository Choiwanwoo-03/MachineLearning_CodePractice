```python
import pandas as pd

age_df = pd.read_csv('3.3.1.age.csv', encoding = 'cp949')
age_df.columns = age_df.columns.str.replace('2025년02월_', '')

name = input('주소 : ')

filtered_df = age_df[age_df['행정구역'].str.contains(name, na = False)]

male_result = filtered_df.iloc[0, 105]
female_result = filtered_df.iloc[0, 208]

male_result = male_result.replace(',', '')
female_result = female_result.replace(',', '')

male_result = int(male_result)
female_result = int(female_result)

size = [male_result, female_result]

color = ['crimson', 'darkcyan']
plt.axis('equal')
plt.pie(size, labels = ['male', 'female'], autopct = '%.1f%%', colors = color, startangle = 90)
plt.legend()
```

```python
import pandas as pd

age_df = pd.read_csv('3.3.1.age.csv', encoding='cp949')
age_df.columns = age_df.columns.str.replace('2025년02월_', '')

# 사용자로부터 원하는 지역 입력 받기
name = input('인구 구조가 알고 싶은 지역의 이름(읍면동 단위)을 입력해주세요 : ')

filtered_df = age_df[age_df['행정구역'].str.contains(name, na=False)]

male_result = filtered_df.iloc[0,106:207].astype(int).values.tolist()
female_result = filtered_df.iloc[0,209:310].astype(int).values.tolist()

plt.scatter(male_result, female_result, c = range(101), alpha = 0.5, cmap = 'jet')
plt.colorbar()
plt.plot(range(max(male_result)), range(max(male_result)), 'g')
plt.xlabel('Male')
plt.ylabel('Female')
```
