## Matplotlib 라이브러리 실습

```python
import matplotlib.pyplot as plt
```

```python
plt.plot([1, 2], [2, 4])
plt.show()
```

```python
y = [2, 3, 4, 5]
plt.plot(y)
plt.show()
```

```python
plt.plot([1, 2, 3, 4], [1, 4, 9, 16])

plt.xlabel("number")
plt.ylabel("square")

plt.show()
```

```python
plt.plot([1, 2, 3, 4], [1, 4, 9, 16], label = 'Square')
plt.xlabel('X-Label')
plt.ylabel('Y-Label')

plt.legend()
plt.show()
```

```python
import numpy as np

x = np.linspace(0, 10, 100)
y1 = x
y2 = 2 * x
y3 = 0.5 * x
y4 = 3 * x

plt.plot(x, y1, label = 'y = x')
plt.plot(x, y2, label = 'y = 2x')
plt.plot(x, y3, label = 'y = 0.5x')
plt.plot(x, y4, label = 'y = 3x')

# plt.legend(loc = 'upper center', ncol = 2)
plt.legend()

plt.title('Simple Linear Plot Example')
plt.xlabel('x')
plt.ylabel('y')
```

```python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3, 4], [3, 6, 9, 12])
plt.xlim([0, 5])
plt.ylim([0, 15])
plt.show()
```

```python
plt.plot([1, 2, 3, 4], [3, 6, 9, 12])

plt.axis([0, 5, 0, 15])
plt.show()
```

```python
x = np.linspace(1, 10, 100)
y = np.exp(x)

plt.subplot(2, 2, 1)
plt.plot(x, y)
plt.title('Linear Scale')
plt.subplot(2, 2, 2)
plt.plot(x, y)
plt.yscale('log')
plt.title('Logarithmic Scale (y-axs)')

plt.subplot(2, 2, 3)
plt.plot(x, y)
plt.xscale('log')
plt.title('Log-Linear Scale (x-axis)')

plt.subplot(2, 2, 4)
plt.plot(x, y)
plt.yscale('log')
plt.xscale('linear')
plt.title('Linear-Logarithmic Scale')

plt.tight_layout()
plt.show()
```

```python
plt.plot([1, 2, 3], [4, 4, 4], '-', label = 'solid')
plt.plot([1, 2,3], [3, 3, 3], '--', label = 'dashed')

plt.plot([1, 2, 3], [2, 2, 2], linestyle = 'dotted', label = 'dotted')
plt.plot([1, 2, 3], [1, 1, 1], linestyle = 'dashdot', label = 'dash-dot')

plt.axis([0.8, 3.2, 0.5, 5.0])
plt.legend(loc = 'upper center', ncol = 4)
plt.show()
```

```python
plt.plot([4, 5, 6], "b")
plt.plot([3, 4, 5], 'ro')
plt.plot([2, 3, 4], marker = "s")
plt.plot([1, 2, 3], marker = "D")
plt.plot([0, 1, 2], marker = "$A$")
plt.show()
```

```python
plt.plot([1, 2, 3, 4], [2.0, 3.3, 6.3, 10.5], color = 'violet')
plt.plot([1, 2, 3, 4], [2.0, 3.1, 5.3, 8.5], color = 'g')
plt.plot([1, 2, 3, 4], [2.0, 2.8, 4.3, 6.3], color = (0.1, 0.2, 0.3))

plt.plot([1, 2, 3, 4], [2.0, 2.5, 3.3, 4.5], color = "#FF0000")
plt.show()
```
