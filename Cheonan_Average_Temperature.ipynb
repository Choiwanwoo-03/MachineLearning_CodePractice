### 1. 파일이름 : 'Cheonan_Temp.csv'

### 1.1 데이터 읽어오기

```python
import pandas as pd

data = pd.read_csv('Cheonan_Temp.csv', encoding = 'cp949')
data
```

```python
# 날짜 앞에 \t 없애기
data['날짜'] = data['날짜'].str.replace('\t', '')
```

```python
data.info()
```

### 0   날짜        문자열
### 1   지점        소수
### 2   평균기온(℃) 소수
### 3   최저기온(℃) 소수
### 4   최고기온(℃) 소수

```python
# 날짜 컬럼의 string 타입을 날짜형식(datetime)으로 변환하기
data['날짜'] = pd.to_datetime(data['날짜'])
data.info()
```

```python
data
```

```python
# 최고기온 날짜와 함께 저장하기
Max_Temp = data.iloc[:, [0, 4]]
Max_Temp
```

```python
# 최저기온 날짜와 함께 저장하기
Min_Temp = data.iloc[:, [0, 3]]
Min_Temp
```

### 1.2 데이터 시각화 하기

```python
import matplotlib.pyplot as plt

plt.plot(Max_Temp['날짜'], Max_Temp['최고기온(℃)'])
plt.xticks(rotation = 45)
plt.title('2025년 최고기온 그래프')

plt.show()
```

```python
# 내가 태어난 달에 해당하는 최고기온 데이터만 추출하기
month_result = Max_Temp[Max_Temp['날짜'].dt.month == 9]
month_result
```

```python
plt.bar(month_result['날짜'], month_result['최고기온(℃)'])
plt.xticks(rotation = 45)
plt.title('9월달 최고기온 그래프')

plt.show()
```

```python
# 내 생일에 해당하는 최저/최고 기온만 추출
birthday_result = Max_Temp[(Max_Temp['날짜'].dt.month == 9) & (Max_Temp['날짜'].dt.day == 29)]
birthday_result

birthday_result2 = Min_Temp[(Min_Temp['날짜'].dt.month == 9) & (Min_Temp['날짜'].dt.day == 29)]
birthday_result2

print(birthday_result)
print(birthday_result2)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. 전체 데이터(data)로 배경 등고선 그리기
# (처음에 불러오신 전체 데이터 변수 이름이 'data'라고 가정합니다)
clean_all_data = data.dropna(subset=['최저기온(℃)', '최고기온(℃)'])

plt.figure(figsize=(10, 8))

# 배경: 천안시 전체 기온의 밀도 분포
sns.kdeplot(
    x=clean_all_data['최저기온(℃)'], 
    y=clean_all_data['최고기온(℃)'], 
    fill=True, 
    cmap='viridis',    # 원하시던 화려한 색상 테마
    alpha=0.8,
    thresh=0.05
)

# 2. 단 하나의 데이터(birthday_result2)를 그 위에 별표로 찍기
plt.scatter(
    birthday_result2['최저기온(℃)'], 
    birthday_result2['최고기온(℃)'], 
    color='red',       # 강렬한 빨간색
    marker='*',        # 별 모양
    s=300,             # 점의 크기 (아주 크게)
    label='My Birthday',
    zorder=5           # 배경에 묻히지 않고 맨 위로 올라오게 설정
)

plt.title('천안시 전체 기온 분포와 특정 날짜(생일)의 기온')
plt.xlabel('최저기온(℃)')
plt.ylabel('최고기온(℃)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```
