### 1.2 데이터 시각화 하기

```python
import matplotlib.pyplot as plt

plt.plot(Max_Temp['날짜'], Max_Temp['최고기온(℃)'])
plt.xticks(rotation = 45)
plt.title('2025년 최고기온 그래프')

plt.show()
```

```python
# 내가 태어난 달에 해당하는 최고기온 데이터만 추출하기
month_result = Max_Temp[Max_Temp['날짜'].dt.month == 9]
month_result
```

```python
plt.bar(month_result['날짜'], month_result['최고기온(℃)'])
plt.xticks(rotation = 45)
plt.title('9월달 최고기온 그래프')

plt.show()
```

```python
# 내 생일에 해당하는 최저/최고 기온만 추출
birthday_result = Max_Temp[(Max_Temp['날짜'].dt.month == 9) & (Max_Temp['날짜'].dt.day == 29)]
birthday_result

birthday_result2 = Min_Temp[(Min_Temp['날짜'].dt.month == 9) & (Min_Temp['날짜'].dt.day == 29)]
birthday_result2

print(birthday_result)
print(birthday_result2)
```

```python
import seaborn as sns

# 1. 전체 데이터로 배경 등고선 그리기
# 최고/최저 기온에 NaN값 제거
cleand_data = data.dropna(subset=['최저기온(℃)', '최고기온(℃)'])

plt.figure(figsize=(10, 8))

# 천안시 전체 기온의 밀도 분포
sns.kdeplot(
    x=cleand_data['최저기온(℃)'], 
    y=cleand_data['최고기온(℃)'], 
    fill=True, 
    cmap='viridis',
    alpha=0.8,
    thresh=0.05
)

# 2. 단 생일 데이터(birthday_result2)를 그 위에 별표로 찍기
plt.scatter(
    birthday_result2['최저기온(℃)'], 
    birthday_result['최고기온(℃)'], 
    color='red',    
    marker='*',   
    s=300,    
    label='My Birthday',
    zorder=5 
)

plt.title('천안시 기온 분포 & 나의 생일 최고/최저기온')
plt.xlabel('최저기온(℃)')
plt.ylabel('최고기온(℃)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

### 3. 그 외 시각화 연습

```python
# 6월과 12월의 최고기온을 fill_between으로 시각화하기
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'NanumGothic'
plt.rcParams['axes.unicode_minus'] = False 

data = pd.read_csv('Cheonan_Temp.csv', encoding='cp949')
data['날짜'] = data['날짜'].str.replace('\t', '')
data['날짜'] = pd.to_datetime(data['날짜'])

cleand_data = data.dropna(subset=['최저기온(℃)', '최고기온(℃)'])
Max_Temp = cleand_data.iloc[:, [0, 4]]

june = Max_Temp[Max_Temp['날짜'].dt.month == 6]
dec = Max_Temp[Max_Temp['날짜'].dt.month == 12]

tempmin_june = june['최고기온(℃)'].min()
tempmin_dec = dec['최고기온(℃)'].min()

fig, ((ax2, ax4)) = plt.subplots(1, 2, figsize = (12, 8))

ax2.fill_between(june['날짜'], tempmin_june, june['최고기온(℃)'], alpha = 0.7, color = 'tomato')
ax2.grid(True)
ax2.label_outer()
ax2.set_title('6월 최고기온')
ax2.tick_params(axis = 'x', rotation = 45)

ax4.fill_between(dec['날짜'], tempmin_dec, dec['최고기온(℃)'], alpha = 0.7, color = 'steelblue')
ax4.grid(True)
ax4.label_outer()
ax4.set_title('12월 최고기온')
ax4.tick_params(axis = 'x', rotation = 45)

fig.suptitle('천안시 6월, 12월 최고기온')
fig.autofmt_xdate()
plt.tight_layout()
plt.show()
```

```python
# 3월과 9월의 최저기온을 violinplot으로 시각화하기
import matplotlib.pyplot as plt
plt.rcParams['font.family'] = 'NanumGothic'
plt.rcParams['axes.unicode_minus'] = False

import pandas as pd
import seaborn as sns

sns.set_theme(style='whitegrid')

data = pd.read_csv('Cheonan_Temp.csv', encoding='cp949')
data['날짜'] = data['날짜'].str.replace('\t', '')
data['날짜'] = pd.to_datetime(data['날짜'])

cleand_data = data.dropna(subset=['최저기온(℃)', '최고기온(℃)'])
Min_Temp = cleand_data.iloc[:, [0, 3]]

march = Min_Temp[Min_Temp['날짜'].dt.month == 3]
sep = Min_Temp[Min_Temp['날짜'].dt.month == 9]

march = march.copy()
march['월'] = '3월'

sep = sep.copy()
sep['월'] = '9월'

df_plot = pd.concat([march, sep])

f, ax = plt.subplots(figsize=(8, 6))

sns.violinplot(data=df_plot, x='월', y='최저기온(℃)',
               bw_adjust=0.5, cut=1, linewidth=1, palette='Set3')

ax.set_title('천안시 3월, 9월 최저기온')
sns.despine(left=True, bottom=True)

plt.tight_layout()
plt.show()
```
