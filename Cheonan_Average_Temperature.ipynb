### 1. 파일이름 : 'Cheonan_Temp.csv'

### 1.1 데이터 읽어오기

```python
import pandas as pd

data = pd.read_csv('Cheonan_Temp.csv', encoding = 'cp949')
data
```

```python
# 년월 앞에 \t 없애기
data['년월'] = data['년월'].str.replace('\t', '')
```

```python
data.info()
```

```python
# 년월 컬럼의 string 타입을 기간(Period)으로 변환하기
data['년월'] = pd.to_datetime(data['년월']).dt.to_period('M')
data.info()
```

```python
data
```

```python
# 평균최고기온만 년월과 함께 저장하기
Max_Temp = data.iloc[:, [0, 4]]
Max_Temp
```

### 1.2 데이터 시각화 하기

```python
import matplotlib.pyplot as plt

plt.plot(Max_Temp['평균최고기온(℃)']) # x축 년월은 생략
```

```python
# 내가 태어난 달에 해당하는 기온 데이터만 추출하기
month_result = Max_Temp[Max_Temp['날짜'].dt.month == 9]
month_result
```
