### 1. 파일이름 : 'Cheonan_Temp.csv'

### 1.1 데이터 읽어오기

```python
import pandas as pd

data = pd.read_csv('Cheonan_Temp.csv', encoding = 'cp949')
data
```

```python
# 날짜 앞에 \t 없애기
data['날짜'] = data['날짜'].str.replace('\t', '')
```

```python
data.info()
```

### 0   날짜        문자열
### 1   지점        소수
### 2   평균기온(℃) 소수
### 3   최저기온(℃) 소수
### 4   최고기온(℃) 소수

```python
# 날짜 컬럼의 string 타입을 날짜형식(datetime)으로 변환하기
data['날짜'] = pd.to_datetime(data['날짜'])
data.info()
```

```python
data
```

```python
# 최고기온만 년월과 함께 저장하기
Max_Temp = data.iloc[:, [0, 4]]
Max_Temp
```

### 1.2 데이터 시각화 하기

```python
import matplotlib.pyplot as plt

plt.plot(Max_Temp['최고기온(℃)']) # x축 년월은 생략
```

```python
# 내가 태어난 달에 해당하는 기온 데이터만 추출하기
month_result = Max_Temp[Max_Temp['날짜'].dt.month == 9]
month_result
```

```python
plt.plot(month_result['최고기온(℃)'])
```

```python
# 내 생일에 해당하는 온도만 추출
birthday_result = Max_Temp[(Max_Temp['날짜'].dt.month == 9) & (Max_Temp['날짜'].dt.day == 29)]
birthday_result
```

```python
plt.plot(birthday_result['최고기온(℃)']) # x축 날짜는 생략
```

```python
birthday_result2 = data[(data['날짜'].dt.month == 9) & (data['날짜'].dt.day == 29)]
birthday_result2
```

```python
plt.plot(birthday_result2['최저기온(℃)'])
plt.plot(birthday_result2['최고기온(℃)'])
```

```python
plt.plot(birthday_result2['날짜'], birthday_result2['최저기온(℃)'])
plt.plot(birthday_result2['날짜'], birthday_result2['최고기온(℃)'])
```

```python
# 한글 폰트 적용
import matplotlib.font_manager as fm

plt.rc('font', family='NanumGothic')
plt.rcParams['axes.unicode_minus'] = False      # - 기호가 깨지는 것을 방지

# 내 생일의 최저 기온과 최고 기온을 모두 추출하기
birthday_result2 = data[(data['날짜'].dt.month == 8) & (data['날짜'].dt.day == 16)]
birthday_result2

plt.plot(birthday_result2['날짜'], birthday_result2['최저기온(℃)'])
plt.plot(birthday_result2['날짜'], birthday_result2['최고기온(℃)'])
plt.title('내 생일의 기온 변화 그래프')
```
