# 2. 라이브러리 및 데이터 불러오기

```python
# 필수 모듈과 데이터 파일 불러오기
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

file_url = 'https://media.githubusercontent.com/media/musthave-ML10/data_source/main/salary.csv'
data = pd.read_csv(file_url, skipinitialspace = True)
```

```python
# 데이터 살펴보기
data = pd.read_csv(file_url, skipinitialspace = False)
data
```

```python
# class 고윳값 및 특징 확인하기
data['class'].unique()
```

```python
data.info()
```

```python
# 통계 정보 출력 - int64 변수만 출력
data.describe()
```

```python
# 통계 정보 출력 - object 형까지 출력
data.describe(include = 'all')
```

# 3.1 전처리 : 범주형 데이터

```python
data['class'] = data['class'].map({'<=50K': 0, '>50K': 1})
data['class']
```

```python
# object 형 변수를 따로 리스트에 저장
obj_list = []
for i in data.columns :
    if data[i].dtype == 'object' :
        obj_list.append(i)
        
obj_list
```

```python
# 각 변수의 고윳값 개수 확인
for i in obj_list :
    print(i, data[i].nunique())
```

```python
# education 과 education-num 변수
for i in np.sort(data['education-num'].unique()) :
    print(i, data[data['education-num'] == i]['education'].unique())
```

```python
# education 변수는 제거
data.drop('education', axis = 1, inplace = True)
```

```python
# occupation 변수 처리
data['occupation'].value_counts()
```

```python
# native-country 변수 처리
data['native-country'].value_counts()
```

```python
# 나라별 class 평균값 확인
data.groupby('native-country').mean(numeric_only = True).sort_values('class')
```

```python
# native-country(범주형)를 국가별 class 평균(숫자)로 치환
data['native-country'] = data['native-country'].map(
    data.groupby('native-country')['class'].mean()
)
data
```

# 3.2 전처리 : 결측치 처리 및 더비 변수 변환

```python
# native-country 결측치 처리
data['native-country'] = data['native-country'].fillna(-99)
```

```python
# workclass(범주형) 결측치 처리
data['workclass'].value_counts()
```

```python
data['workclass'] = data['workclass'].fillna('Private')
```

```python
# occupation(범주형) 결측치 처리
data['occupation'].value_counts()
```

```python
data['occupation'] = data['occupation'].fillna('Unknown')
```

```python
# 나머지 범주형 변수를 더미 변수로 변환
data = pd.get_dummies(data, drop_first = True)
```

# 4.1 모델링 및 평가하기

```python
# 훈련셋과 시험셋 나누기
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(data.drop('class', axis = 1), data['class'],
                                                    test_size = 0.4, random_state = 100)
```

```python
# 결정트리 모델 선택하기
from sklearn.tree import DecisionTreeClassifier
```

```python
# 모델링 및 예측 하기
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
pred = model.predict(X_test)
```

```python
# 정확도 확인하기
from sklearn.metrics import accuracy_score
accuracy_score(y_test, pred)
```

# 7. 매개변수 튜닝

```python
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
train_pred = model.predict(X_train)
test_pred = model.predict(X_test)
print("Train score : ", accuracy_score(y_train, train_pred), end = ' ')
print("Test score : ", accuracy_score(y_test, test_pred))
```
