```python
import pandas as pd

data = pd.read_csv('3.1.2.temp.csv', encoding = 'cp949')
data['날짜'] = data['날짜'].str.replace('\t', '')
data['날짜'] = pd.to_datetime(data['날짜'])

result = data.iloc[:, [0, 4]]

plt.hist(result['최고기온(℃)'], bins = 100, color = 'r')
```

```python
month_result = result[result['날짜'].dt.month == 8]

plt.hist(month_result['최고기온(℃)'], bins = 100, color = 'r')
```

```python
aug_result = result[result['날짜'].dt.month == 8]
jan_result = result[result['날짜'].dt.month == 1]

plt.hist(aug_result['최고기온(℃)'], bins = 100, color = 'r', label = 'Aug')
plt.hist(jan_result['최고기온(℃)'], bins = 100, color = 'b', label = 'Jan')
```

```python
data = pd.read_csv('3.1.2.temp.csv', encoding = 'cp949')
data['날짜'] = data['날짜'].str.replace('\t', '')
data['날짜'] = pd.to_datetime(data['날짜'])

result = data.iloc[:, [0, 4]]

monthly_temp = []
for month in range(1, 13) :
    monthly_temp.append(list(result[result['날짜'].dt.month == month]['최고기온(℃)']))
    
print(monthly_temp)

plt.boxplot(monthly_temp)
```
